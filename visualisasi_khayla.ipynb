```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load dataset langsung dari URL GitHub Anda
url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv"
df = pd.read_csv(url)

# 2. Mengambil HANYA 10 data pertama
df_top10 = df.head(10)

print("✨ 10 Data Pertama dari Dataset Anda:")
print(df_top10)

# 3. Mengatur tema visual pastel yang imut
sns.set_theme(style="whitegrid", palette="pastel")

# Memilih dua kolom numerik pertama
numeric_cols = df_top10.select_dtypes(include=["number"]).columns

if len(numeric_cols) >= 2:
  col_x = numeric_cols[0]
  col_y = numeric_cols[1]

  # 4. Membuat Kanvas dengan Warna Pastel yang Lembut (LavenderBlush)
  plt.figure(figsize=(10, 6), facecolor="#FFF0F5")
  ax = plt.axes()
  ax.set_facecolor("#FFFBF0")  # Warna latar dalam plot (Warm Cream)

  # 5. Membuat Bar Plot / Line Plot yang imut untuk 10 data pertama
  # Menggunakan kombinasi garis dan titik (Marker)
  plt.plot(
      df_top10.index,
      df_top10[col_y],
      marker="o",
      markersize=12,
      color="#FF69B4",  # Hot Pink
      linewidth=3,
      linestyle="-",
      markerfacecolor="#FFB6C1",
      markeredgecolor="#8B008B",
      markeredgewidth=1.5,
      label=f"Nilai {col_y}",
  )

  # 6. Menambahkan Judul & Label dengan Sentuhan Ceria
  plt.title(
      f"✨ 10 Data Pertama: Analisis Ceria ({col_y}) ✨\n(Dataset Fiskal 97"
      " HipoFull)",
      fontsize=14,
      fontweight="bold",
      color="#C71585",
      pad=20,
  )
  plt.xlabel(
      "🌱 Urutan Data ke- (1 sampai 10)",
      fontsize=12,
      color="#4B0082",
      fontweight="bold",
  )
  plt.ylabel(
      f"🚀 Nilai {col_y}", fontsize=12, color="#4B0082", fontweight="bold"
  )

  # Menyesuaikan label sumbu X agar menampilkan indeks 0-9 atau 1-10
  plt.xticks(
      df_top10.index, [f"Data {i+1}" for i in range(len(df_top10))], rotation=30
  )

  # 7. Memberikan Anotasi Imut di Titik Tertinggi dari 10 Data Tersebut
  max_idx = df_top10[col_y].idxmax()
  max_val = df_top10.loc[max_idx, col_y]

  plt.annotate(
      "🌟 Bintang Utama (Tertinggi)",
      xy=(max_idx, max_val),
      xytext=(max_idx, max_val * 1.08),
      arrowprops=dict(
          facecolor="#FF1493",
          shrink=0.08,
          width=2,
          headwidth=8,
          connectionstyle="arc3,rad=.2",
      ),
      fontsize=10,
      fontweight="bold",
      color="#8B008B",
      bbox=dict(boxstyle="round,pad=0.4", fc="#FFE4E1", ec="#FF69B4", lw=1.5),
  )

  # Memperindah grid
  plt.grid(True, linestyle=":", alpha=0.6, color="#D8BFD8")
  plt.legend(loc="upper left", facecolor="#FFF0F5", edgecolor="#FF69B4")
  plt.tight_layout()

  # Tampilkan grafik
  plt.show()
else:
  print("⚠️ Kolom numerik kurang dari 2.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import pandas as pd

# Data (Plotly lebih suka data dalam bentuk DataFrame)
df = pd.DataFrame({
    'Bulan': ['Jan', 'Feb', 'Mar', 'Apr', 'Mei'],
    'Penjualan': [150, 200, 180, 250, 300]
})

# A. Dictionary
# B. Tuple
# C. List

# Membuat grafik dasar
fig = px.bar(df, x='Bulan', y='Penjualan', title='Penjualan per Bulan')

fig.update_layout(width=700, height=400) #pixel

# Menampilkan grafik
fig.show()
bulan = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Juni', 'Juli']
penjualan = [150, 200, 180, 250, 125, 200, 130]

plt.figure(figsize=(8, 4))

sns.barplot(x=bulan, y=penjualan, palette="muted")

plt.plot(
    range(len(bulan)),
    penjualan,
    color='red',
    marker='o',
    linestyle='-',
    linewidth=2,
    label='Garis Tren'
)

plt.title('Penjualan per Bulan dengan Garis Tren', fontsize=12, fontweight='bold')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Penjualan')
plt.legend()

plt.tight_layout()
plt.show()
```
